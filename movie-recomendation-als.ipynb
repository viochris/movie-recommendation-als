# 🎬 Movie Recommendation System — ALS Collaborative Filtering on MovieLens (via Apache Spark)

This notebook builds a **collaborative filtering movie recommender** using **Alternating Least Squares (ALS)** matrix factorization, implemented with **Apache Spark's MLlib** (`pyspark.ml.recommendation.ALS`), trained on the same **MovieLens "latest small"** dataset used in this project's SVD companion notebook (100,836 ratings from 610 users across 9,721 rated movies after cleaning).

This notebook shares the same data foundation as the SVD version (loading, duplicate-ID cleaning, EDA), then diverges at the modeling stage: instead of a single-machine `scikit-surprise` model, everything from data conversion onward runs through a Spark session.

**What this notebook covers, end to end:**
1. Loading and validating the four raw MovieLens CSV files (`movies`, `ratings`, `tags`, `links`).
2. Auditing and resolving a handful of duplicate movie IDs against external IMDb/TMDb identifiers.
3. Exploratory Data Analysis (EDA) on rating and user behavior patterns.
4. Initializing a Spark session and converting the cleaned data into a Spark DataFrame.
5. Training a baseline ALS model and evaluating it with 5-fold cross-validation.
6. Tuning hyperparameters with Spark's `CrossValidator` and `ParamGridBuilder`.
7. Refitting the tuned model on the full dataset for production use.
8. Generating personalized top-10 recommendations for a target user.
9. Saving the trained model to disk using Spark's native model writer.

## 3. Data Cleaning — Duplicate Movie ID Resolution

Before merging or training on this data, it's worth checking whether any movie title maps to more than one `movieId`. If the same film was accidentally recorded under two different IDs during data collection, that would split its rating signal in two and understate its true popularity to any model trained on it. The cell below flags every title that has more than one associated ID.

```python
# Aggregation to isolate duplicate entries by counting unique IDs per movie title
id_counts_by_title = movies_df.groupby("title")["movieId"].nunique().reset_index(name="id_count")

# Filters explicitly for titles mapped to multiple distinct IDs
duplicate_titles = id_counts_by_title[id_counts_by_title["id_count"] > 1].sort_values(by="id_count", ascending=False)

# Terminal UI header for duplicate detection results
print("\n" + "=" * 70)
print(" TITLES WITH MULTIPLE MOVIE IDs ".center(70, "="))
print("=" * 70)

# Conditional display handling based on anomaly presence
if len(duplicate_titles) > 0:
    display(duplicate_titles)
else:
    print("✅ [CLEAN] No duplicate movie titles found across the dataset.")
print("-" * 70 + "\n")
```

### 3.1 Cross-Reference Duplicates Against External IDs

A duplicated title alone isn't proof of a data error, since two genuinely different films (a remake, a different adaptation) can share the exact same title. To tell the two cases apart, each duplicated title's records are joined against `links_df`, which carries their external IMDb and TMDb identifiers. Comparing genres and external IDs side by side makes it possible to judge, case by case, whether a duplicate is a real linking error or a legitimately distinct movie.

```python
# Extraction of all duplicate title records to verify consistency across external databases (IMDb/TMDb)
duplicate_title_records = movies_df[movies_df.duplicated("title", keep=False)].sort_values(by="title")

# Integration with external link identifiers for deeper anomaly resolution
merged_duplicate_data = duplicate_title_records.merge(links_df, on="movieId", how="left")

# Terminal UI header for cross-reference anomaly detection results
print("\n" + "=" * 70)
print(" DUPLICATE TITLES & EXTERNAL LINKS MATCHING ".center(70, "="))
print("=" * 70)

# Conditional display handling based on anomaly presence
if len(merged_duplicate_data) > 0:
    print(f"Total anomalous records to review : {len(merged_duplicate_data)}")
    print("-" * 70)
    display(merged_duplicate_data[['movieId', 'title', 'genres', 'imdbId', 'tmdbId']])
else:
    print("✅ [CLEAN] No duplicate movie titles require external cross-referencing.")
print("-" * 70 + "\n")
```

### 3.2 Resolve and Consolidate Duplicate IDs

Based on the manual review above, five duplicated titles were found in total. Three of them (*Confessions of a Dangerous Mind*, *Eros*, *Saturn 3*) have consistent genres across their duplicate rows, and their missing `tmdbId` values turned out to simply be a linking failure during data collection rather than a sign of a different film, so their IDs are safely merged into one canonical ID each. The other two (*Emma*, *War of the Worlds*) have conflicting genres and distinct TMDb IDs, confirming they're genuinely different productions sharing a title (for example, a theatrical release versus a TV movie), so they are deliberately **left unmerged** to avoid corrupting the data. The mapping below is applied across `ratings_df`, `tags_df`, and `movies_df` so every dataset stays consistent.

```python
# =========================================================
# DUPLICATE ID CONSOLIDATION
# =========================================================

# CRITICAL DATA VALIDATION: 
# Anomalies are resolved based on genre consistency and external ID matching.
# 
# 1. EXCLUDED FROM MERGING (Distinct Productions):
#    - "Emma" & "War of the Worlds": Despite sharing titles and release years, they have 
#      conflicting genres and distinct TMDB IDs. This confirms they are separate adaptations 
#      (e.g., theatrical vs. TV movie, or blockbuster vs. indie). Merging would corrupt data.
#
# 2. INCLUDED IN MERGING (Genuine Data-Linking Errors):
#    - "Confessions of a Dangerous Mind", "Eros", & "Saturn 3": Genres are highly consistent. 
#      The presence of NaN in tmdbId for some of these rows is simply a failure to link 
#      to the external TMDB database during data collection, not an indicator of a different film. 
#      These are safe to consolidate.

# Dictionary mapping obsolete or duplicate movie IDs to their primary canonical IDs
duplicate_movie_mapping = {
    6003: 144606,    # Confessions of a Dangerous Mind
    # 26958: 838,    # Emma (DO NOT MERGE - Distinct adaptations with conflicting genres/TMDB IDs)
    32600: 147002,   # Eros (SAFE TO MERGE - NaN tmdbId is just a linking failure; genres match)
    168358: 2851,    # Saturn 3 (SAFE TO MERGE - NaN tmdbId is just a linking failure; genres match)
    # 64997: 34048   # War of the Worlds (DO NOT MERGE - Distinct adaptations with conflicting genres/TMDB IDs)
}

# Terminal UI header for the data consolidation process
print("\n" + "=" * 70)
print(" RESOLVING DUPLICATE MOVIE IDs ".center(70, "="))
print("=" * 70)

# Standardize foreign keys in the relational datasets using the canonical mapping
ratings_df["movieId"] = ratings_df["movieId"].replace(duplicate_movie_mapping)
tags_df["movieId"] = tags_df["movieId"].replace(duplicate_movie_mapping)

# Standardize the primary key in the movies dataset and drop redundant title records
movies_df["movieId"] = movies_df["movieId"].replace(duplicate_movie_mapping)
movies_df = movies_df.drop_duplicates(subset=["movieId"], keep="first")

# Execution confirmation output
print(f"✅ [SUCCESS] Consolidated {len(duplicate_movie_mapping)} duplicate titles across all datasets.")
print("-" * 70 + "\n")
```

## 6. Set Up Apache Spark and Convert the Data

### 6.1 Initialize a Spark Session

Since the modeling stage runs on Spark MLlib rather than a single-machine library, a `SparkSession` is created first. Two extra settings are configured here: Apache Arrow is enabled for faster, more efficient conversion between pandas and Spark DataFrames, and a checkpoint directory is set up specifically because ALS's iterative training process can otherwise cause a `StackOverflowError` from Spark's lazy execution lineage growing too deep without periodic checkpointing.

```python
# Terminal UI header for Spark Session
print("\n" + "=" * 70)
print(" SPARK SESSION INITIALIZATION ".center(70, "="))
print("=" * 70)

# Initialize a Spark session
print("Initializing Spark session for distributed processing")
spark = SparkSession.builder.appName("MovieRecommendationALS").getOrCreate()
print("✅ SUCCESS Spark session initialized")

# Enable Arrow based columnar data transfers for faster conversion
print("\nEnabling Apache Arrow optimization")
spark.conf.set("spark.sql.execution.arrow.pyspark.enabled", "true")
print("✅ SUCCESS Apache Arrow optimization enabled")

# Configure global checkpointing directory to prevent StackOverflow errors in ALS
print("\nConfiguring global checkpointing directory")
checkpoint_dir = "output/checkpoints"
os.makedirs(checkpoint_dir, exist_ok=True)
spark.sparkContext.setCheckpointDir(checkpoint_dir)
print(f"✅ SUCCESS Checkpoint directory set to '{checkpoint_dir}'\n")

print("=" * 70 + "\n")
```

### 6.2 Convert `combined_df` into a Spark DataFrame

Only the three columns ALS actually needs (`userId`, `movieId`, `rating`) are converted from the pandas `combined_df` into a Spark DataFrame, `ratings_spark`, which is what the rest of the modeling pipeline will operate on. The schema and a preview of the first 5 rows are printed to confirm the conversion worked as expected.

```python
# Terminal UI header for Spark DataFrame creation
print("\n" + "=" * 70)
print(" SPARK DATAFRAME CREATION AND PREVIEW ".center(70, "="))
print("=" * 70)

# Convert the required pandas DataFrame columns into a Spark DataFrame
print("Converting pandas DataFrame to Spark DataFrame")
ratings_spark = spark.createDataFrame(combined_df[["userId", "movieId", "rating"]])
print("✅ SUCCESS Conversion successful\n")

print("DataFrame Schema")
ratings_spark.printSchema()

print("Data Preview Top 5 Rows")
ratings_spark.show(5)
print("=" * 70 + "\n")
```

### 6.3 Train-Test Split

`ratings_spark` is split 80/20 into a training set and a held-out test set using Spark's own `randomSplit()`, seeded with the same `random_state` used throughout the notebook for reproducibility. The training set is what the model will actually learn from; the test set is kept aside purely for evaluating how well the model generalizes to ratings it has never seen.

```python
# Terminal UI header for the data splitting process
print("\n" + "=" * 70)
print(" TRAIN TEST SPLIT ".center(70, "="))
print("=" * 70)
print("Splitting the Spark DataFrame into training 80 and testing 20 sets")

# Perform the random split using the predefined random state for reproducibility
training, test = ratings_spark.randomSplit([0.8, 0.2], seed=random_state)

# Calculate and display the row counts for each dataset
train_count = training.count()
test_count = test.count()

print(f"Total Training Rows {train_count}")
print(f"Total Testing Rows  {test_count}")
print("=" * 70 + "\n")
```

## 7. Baseline ALS Model

### 7.1 Initialize the Model

A baseline `ALS` model is configured with `rank=100`, `maxIter=20`, and `regParam=0.1` before any tuning happens, so its performance can later be compared against a tuned version. A few other settings are worth noting: `coldStartStrategy="drop"` tells Spark to drop any rows in evaluation data where the user or movie wasn't seen during training (rather than producing a `NaN` prediction that would break the evaluation metrics), and `nonnegative=False` allows the model's learned latent factors to take negative values, which is the standard, unconstrained version of matrix factorization.

```python
# Terminal UI header for the ALS model initialization
print("\n" + "=" * 70)
print(" ALS MODEL INITIALIZATION ".center(70, "="))
print("=" * 70)
print("Configuring the Alternating Least Squares ALS model parameters")

# Initialize the ALS model with specified hyperparameters
model = ALS(
    userCol="userId",
    itemCol="movieId",
    ratingCol="rating",
    rank=100,
    maxIter=20,
    regParam=0.1,
    coldStartStrategy="drop",
    nonnegative=False,
    seed=random_state
)

# Set checkpoint interval on the model (checkpoint directory is set globally)
model.setCheckpointInterval(2)

# Execution confirmation
print("✅ SUCCESS ALS model configured successfully\n")
print("=" * 70 + "\n")
```

### 7.2 5-Fold Cross-Validation (Single Configuration)

Before comparing against any tuned alternative, the baseline configuration itself is evaluated with 5-fold cross-validation using Spark's `CrossValidator`. Since `ParamGridBuilder` is given only one value for each hyperparameter here, this isn't really a search yet, it's a single configuration being cross-validated to get a robust RMSE estimate for the baseline before moving on to an actual grid search.

```python
# Terminal UI header for the cross validation process
print("\n" + "=" * 70)
print(" 5 FOLD CROSS VALIDATION SINGLE PARAMETER GRID ".center(70, "="))
print("=" * 70)

# Define a single parameter grid to evaluate a specific model configuration
single_param_grid = ParamGridBuilder() \
    .addGrid(model.rank, [100]) \
    .addGrid(model.maxIter, [20]) \
    .addGrid(model.regParam, [0.1]) \
    .build()

print("Initiating 5 fold cross validation This may take a while\n")

# Initialize evaluator specifically for cross validation optimizing for RMSE
evaluator_rmse = RegressionEvaluator(metricName="rmse", labelCol="rating", predictionCol="prediction")

# Configure the CrossValidator
cross_val = CrossValidator(
    estimator=model,
    estimatorParamMaps=single_param_grid,
    evaluator=evaluator_rmse,
    numFolds=5
)

# Execute the cross validation process with execution time tracking
start_time = time.time()
cv_model = cross_val.fit(training)
end_time = time.time()

# Time conversion for human readable output
elapsed_time = end_time - start_time
minutes = int(elapsed_time // 60)
seconds = elapsed_time % 60

# Execution confirmation and formatted output
print(f"✅ SUCCESS Cross validation completed in {minutes}m {seconds:.2f}s\n")

# Extract and display the average cross validation metric
cv_rmse = cv_model.avgMetrics[0]

# Terminal UI sub header strictly for the evaluation result
print("=" * 70)
print(" CROSS VALIDATION RESULT ".center(70, "="))
print("=" * 70)
print(f"Average RMSE Score           {cv_rmse:.4f}")
print("=" * 70 + "\n")
```

### 7.3 Fit the Baseline Model on the Training Split

With the baseline configuration's cross-validated performance established, the model is now fit directly on the `training` DataFrame created in Section 6.3, so it can be evaluated on the held-out test set next.

```python
# Terminal UI header for the ALS model fitting process
print("\n" + "=" * 70)
print(" ALS MODEL FITTING ".center(70, "="))
print("=" * 70)
print("Fitting the ALS model on the training data This may take a while")

# Execute the fitting process with execution time tracking
start_time = time.time()
fitted_als = model.fit(training)
end_time = time.time()

# Time conversion for human readable output
elapsed_time = end_time - start_time
minutes = int(elapsed_time // 60)
seconds = elapsed_time % 60

# Execution confirmation and formatted output
print(f"✅ SUCCESS Model fitted successfully in {minutes}m {seconds:.2f}s\n")
print("=" * 70 + "\n")
```

### 7.4 Evaluate the Baseline Model on the Held-Out Test Set

The trained baseline model generates predictions for the test set, and `RegressionEvaluator` is used to compute RMSE, MAE, and R² by comparing those predictions against the true ratings on data the model never saw during training.

```python
# Terminal UI header for model evaluation
print("\n" + "=" * 70)
print(" ALS MODEL EVALUATION TEST SET ".center(70, "="))
print("=" * 70)
print("Generating predictions on the testing data")

# Generate predictions
predictions = fitted_als.transform(test)

print("\nPredictions Schema")
predictions.printSchema()

# ==============================================================================
# EVALUATION METRICS CALCULATION
# 
# Note There are 3 common ways to evaluate multiple metrics in PySpark
#
# METHOD 1 Instantiating separate evaluators for each metric
# evaluator_rmse = RegressionEvaluator(metricName="rmse", labelCol="rating", predictionCol="prediction")
# rmse = evaluator_rmse.evaluate(predictions)
#
# METHOD 2 Passing a parameter map directly to the evaluate function
# evaluator = RegressionEvaluator(labelCol="rating", predictionCol="prediction")
# rmse = evaluator.evaluate(predictions, {evaluator.metricName: "rmse"})
#
# METHOD 3 Using setter methods on a single evaluator instance Active choice below
# evaluator = RegressionEvaluator(labelCol="rating", predictionCol="prediction")
# evaluator.setMetricName("rmse")
# rmse = evaluator.evaluate(predictions)
# ==============================================================================

print("Calculating RMSE MAE and R2 metrics")

# Initialize a single evaluator instance
evaluator = RegressionEvaluator(labelCol="rating", predictionCol="prediction")

# Calculate metrics using Method 3
evaluator.setMetricName("rmse")
rmse = evaluator.evaluate(predictions)

evaluator.setMetricName("mae")
mae = evaluator.evaluate(predictions)

evaluator.setMetricName("r2")
r2 = evaluator.evaluate(predictions)

# Terminal UI sub header strictly for model evaluation metrics
print("\n" + "=" * 70)
print(" FINAL EVALUATION METRICS TEST SET ".center(70, "="))
print("=" * 70)
print(f"Root Mean Squared Error RMSE {rmse:.4f}")
print(f"Mean Absolute Error MAE      {mae:.4f}")
print(f"R Squared R2                 {r2:.4f}")
print("=" * 70 + "\n")
```

## 8. Hyperparameter Tuning with CrossValidator

Rather than assuming the baseline hyperparameters are optimal, a proper hyperparameter search is run this time: `ParamGridBuilder` is configured with 8 combinations of `rank` (50, 100) and `maxIter` (15, 20) and `regParam` (0.05, 0.1), each evaluated with 5-fold cross-validation. `parallelism=4` lets Spark evaluate multiple parameter combinations at the same time rather than strictly one after another, which helps offset how much slower this step is compared to the single-configuration cross-validation above.

```python
# Terminal UI header for hyperparameter tuning
print("\n" + "=" * 70)
print(" HYPERPARAMETER TUNING CROSS VALIDATION ".center(70, "="))
print("=" * 70)

# Define the hyperparameter grid for ALS model (8 combinations)
param_grid = ParamGridBuilder() \
    .addGrid(model.rank, [50, 100]) \
    .addGrid(model.maxIter, [15, 20]) \
    .addGrid(model.regParam, [0.05, 0.1]) \
    .build()

print(f"Total parameter combinations to test {len(param_grid)}")
print("Initiating 5 fold cross validation with parallel execution\n")

# Initialize evaluator specifically for cross validation optimizing for RMSE
evaluator_rmse = RegressionEvaluator(metricName="rmse", labelCol="rating", predictionCol="prediction")

# Configure the CrossValidator
cross_validator = CrossValidator(
    estimator=model,
    estimatorParamMaps=param_grid,
    evaluator=evaluator_rmse,
    numFolds=5,
    parallelism=4
)

# Execute the cross validation process with execution time tracking
start_time = time.time()
cv_model = cross_validator.fit(training)
end_time = time.time()

# Time conversion for human readable output
elapsed_time = end_time - start_time
minutes = int(elapsed_time // 60)
seconds = elapsed_time % 60

print(f"✅ SUCCESS Grid search completed in {minutes}m {seconds:.2f}s\n")

# Extract best metrics and parameters
best_score = min(cv_model.avgMetrics)
best_index = cv_model.avgMetrics.index(best_score)
best_params = param_grid[best_index]

# Terminal UI sub header strictly for the best model parameters
print("=" * 70)
print(" OPTIMAL HYPERPARAMETERS FOUND ".center(70, "="))
print("=" * 70)
print(f"Target Metric RMSE Best Score {best_score:.4f}\n")

print("Best Parameters from ParamGridBuilder")
for param, value in best_params.items():
    print(f"  - {param.name:<20} : {value}")

print("\nBest Parameters from extractParamMap")
for param, value in cv_model.bestModel.extractParamMap().items():
    print(f"  - {param.name:<20} : {value}")
print("=" * 70 + "\n")
```

### 8.1 Evaluate the Best Model on the Held-Out Test Set

`cv_model.bestModel` is the model already refit on the training data with the winning hyperparameter combination from the search above, so it can be evaluated on the same held-out test set used for the baseline, giving a fair, apples-to-apples comparison between the two.

```python
# Terminal UI header for optimal model evaluation
print("\n" + "=" * 70)
print(" OPTIMIZED ALS MODEL EVALUATION TEST SET ".center(70, "="))
print("=" * 70)
print("Generating predictions using the best estimator on the testing data")

# Extract the optimal model and generate predictions
best_estimator = cv_model.bestModel
predictions = best_estimator.transform(test)

print("Calculating RMSE MAE and R2 metrics using parameter map")

# Initialize a single evaluator instance
evaluator = RegressionEvaluator(labelCol="rating", predictionCol="prediction")

# Calculate metrics by passing a parameter map Method 2
rmse = evaluator.evaluate(predictions, {evaluator.metricName: "rmse"})
mae = evaluator.evaluate(predictions, {evaluator.metricName: "mae"})
r2 = evaluator.evaluate(predictions, {evaluator.metricName: "r2"})

# Terminal UI sub header strictly for optimized model evaluation metrics
print("\n" + "=" * 70)
print(" FINAL OPTIMIZED EVALUATION METRICS TEST SET ".center(70, "="))
print("=" * 70)
print(f"Root Mean Squared Error RMSE {rmse:.4f}")
print(f"Mean Absolute Error MAE      {mae:.4f}")
print(f"R Squared R2                 {r2:.4f}")
print("=" * 70 + "\n")
```

## 9. Final Model — Train on the Full Dataset

For actual production use, discarding 20% of the ratings as a test set would mean throwing away real signal the model could otherwise learn from. The winning hyperparameters found by the search above are extracted here and used to configure a fresh `ALS` instance, `final_model`, which is then fit on `ratings_spark` in its entirety rather than just the 80% training split.

```python
# Terminal UI header for the final model deployment
print("\n" + "=" * 70)
print(" FINAL OPTIMIZED ALS MODEL DEPLOYMENT ".center(70, "="))
print("=" * 70)
print("Extracting optimal hyperparameters and fitting on the entire dataset\n")

# ==============================================================================
# PARAMETER EXTRACTION NOTES
# best_params is a dictionary where KEYS are Param objects, not strings.
# Example structure:
# {
#     Param(parent='ALS_xxx', name='rank', doc='...'): 100,     # 'name' is fixed
#     Param(parent='ALS_xxx', name='maxIter', doc='...'): 20,   # 'name' is fixed
#     Param(parent='ALS_xxx', name='regParam', doc='...'): 0.1  # 'name' is fixed
# }
# NOTE: 'parent' is a random auto-generated ID. 
# NOTE: 'doc' is the parameter's docstring text.
# 
# Extracting values uses the exact identical Param object from the model:
# ==============================================================================

best_rank = best_params[model.rank]         # Evaluates to 100
best_maxIter = best_params[model.maxIter]   # Evaluates to 20
best_regParam = best_params[model.regParam] # Evaluates to 0.1

# Initialize the final ALS model with the optimal hyperparameters
final_model = ALS(
    userCol="userId",
    itemCol="movieId",
    ratingCol="rating",
    rank=best_rank,
    maxIter=best_maxIter,
    regParam=best_regParam,
    coldStartStrategy="drop",
    nonnegative=False,
    seed=random_state
)

# Set checkpoint interval on the new final model instance (dir is set globally)
final_model.setCheckpointInterval(2)

print("Initiating final model training This may take a while")

# Execute the fitting process with execution time tracking
start_time = time.time()
fitted_final_model = final_model.fit(ratings_spark)
end_time = time.time()

# Time conversion for human readable output
elapsed_time = end_time - start_time
minutes = int(elapsed_time // 60)
seconds = elapsed_time % 60

# Execution confirmation and formatted output
print(f"✅ SUCCESS Final model fitted successfully in {minutes}m {seconds:.2f}s\n")
print("=" * 70 + "\n")
```

## 10. Generate Recommendations

### 10.1 Define `get_top_n_recommendations()`

This helper function wraps Spark's built-in `recommendForUserSubset()` method: it builds a single-row Spark DataFrame for the target user, asks the fitted ALS model for its top N recommendations, then unpacks the nested array-of-structs result into a flat pandas DataFrame. Predicted ratings are clipped to the valid 0.5–5.0 range (kept alongside the unclipped `raw_predicted_rating` for reference), and the result is merged with `movies_df` to attach readable titles and genres.

```python
# Terminal UI header for the recommendation function definition
print("\n" + "=" * 70)
print(" RECOMMENDATION ENGINE FUNCTION ".center(70, "="))
print("=" * 70)
print("Defining function to extract and format top N recommendations")

def get_top_n_recommendations(model, movies_df, user_id, n=10):
    try:
        # Create a Spark DataFrame for the specific user (requires tuple format)
        user_df = spark.createDataFrame([(user_id,)], schema=["userId"])
        # user_df:
        # +------+
        # |userId|
        # +------+
        # |     5|
        # +------+

        # Generate recommendations using Spark ALS's built-in method
        recommendations = model.recommendForUserSubset(user_df, n)
        # recommendations (Spark DataFrame, 1 row per user, recs packed into 1 array column):
        # +------+-----------------------------------------------------------+
        # |userId|recommendations                                            |
        # +------+-----------------------------------------------------------+
        # |     5|[{318, 4.8}, {296, 4.7}, ...]  <- array of structs         |
        # +------+-----------------------------------------------------------+

        # Convert to pandas for easier manipulation of the nested structure
        recs_pandas = recommendations.toPandas()
        # recs_pandas (same shape as above, just now a pandas DataFrame):
        #    userId                                                recommendations
        # 0       5  [{'movieId': 318, 'rating': 4.8}, {'movieId': 296, 'rating': 4.7}]

        # Guard clause for empty recommendations
        if recs_pandas.empty:
            print(f"No recommendations found for user {user_id}")
            return pd.DataFrame()

        # Explode the array of recommendation structs into individual rows
        recs_exploded = recs_pandas.explode("recommendations")
        # recs_exploded (1 row per movie now, instead of 1 row per user):
        #    userId                  recommendations
        # 0       5  {'movieId': 318, 'rating': 4.8}
        # 0       5  {'movieId': 296, 'rating': 4.7}

        # Extract movieId and rating from the struct into their own columns
        recs_exploded["movieId"] = recs_exploded["recommendations"].apply(lambda x: x["movieId"])
        recs_exploded["raw_predicted_rating"] = recs_exploded["recommendations"].apply(lambda x: x["rating"])
        recs_exploded["predicted_rating"] = recs_exploded["raw_predicted_rating"].clip(lower=0.5, upper=5.0)
        # recs_exploded (struct unpacked into flat columns):
        #    userId  ...  movieId  predicted_rating
        # 0       5  ...      318               4.8
        # 0       5  ...      296               4.7

        # Merge with movie metadata to get titles and genres
        final_recs = recs_exploded.merge(movies_df, on="movieId", how="left")
        # final_recs (now joined with title/genres from movies_df):
        #    movieId                 title                genres  predicted_rating
        # 0      318  Shawshank Redemption           Crime|Drama               4.8
        # 1      296          Pulp Fiction  Crime|Drama|Thriller               4.7

        # Return the cleaned up and filtered DataFrame
        return final_recs[["movieId", "title", "genres", "raw_predicted_rating", "predicted_rating"]]

    except Exception as e:
        print(f"Error generating recommendations {e}")
        return pd.DataFrame()

print("✅ SUCCESS Function get_top_n_recommendations defined successfully\n")
print("=" * 70 + "\n")
```

This notebook set out to solve the same underlying problem as its SVD companion notebook (a large, unpersonalized catalog costing engagement, and a sparse, noisy rating dataset that makes a naive similarity approach unreliable), while also testing a different technical angle: whether the same collaborative filtering approach can be trained on a distributed engine instead of a single machine.

The ALS pipeline built here addresses all three of the problems it set out to solve. It learns latent taste factors purely from the rating matrix via Spark MLlib's ALS implementation, letting it rank the entire unwatched catalog for any given user. And it trains through Apache Spark rather than a single-process library, so the same code path is, in principle, ready to be pointed at a cluster and a catalog far larger than this one without a fundamental redesign.

That said, this is a solution to those specific problems, not a complete, production-ready recommender, and the sections below go through exactly where it still falls short.

## 13. System Limitations

**Architectural Limitations**
- **No serving layer.** Recommendations are only produced by calling `get_top_n_recommendations()` inside this notebook. There's no API endpoint or UI to query the trained model interactively without opening and rerunning cells.
- **Distributed overhead with no distributed benefit, yet.** Spark runs here in local mode on a single machine, so it pays for Spark's job scheduling, serialization, and checkpointing overhead without actually gaining a speed advantage from distribution. The scalability argument for using Spark is architectural and forward-looking rather than something this specific run demonstrates, since a genuinely large, multi-node cluster was never used here.
- **Single held-out split for the final numbers.** The final model's headline metrics come from one 80/20 train-test split layered on top of the 5-fold cross-validation used during grid search, so those specific numbers carry a bit more split-to-split variance than the cross-validated averages do.
- **Unused imports left over from a shared template.** The setup cell imports a wide range of classification and tuning libraries (XGBoost, LightGBM, CatBoost, `imbalanced-learn`, LIME, `joblib`, and Optuna) that are never actually called anywhere in this notebook's ALS pipeline. `joblib` in particular is worth calling out since the final model is saved through Spark's own writer, not `joblib.dump()`.
- **A duplicated EDA cell.** As noted earlier, the rating distribution histogram appears twice in a row with identical code and output, which looks like a leftover copy-paste rather than intentional.

**Model & Domain Limitations**
- **Structural cold-start problem.** ALS here learns purely from `(userId, movieId, rating)` triples. It has no way to score a brand-new movie that has zero ratings yet, or a brand-new user with no rating history at all, because there's no learned latent vector for either one to fall back on. `genres.csv` and `tags.csv` are loaded and used for display and duplicate-ID resolution, but never as model input, so none of that metadata currently helps with this gap. `coldStartStrategy="drop"` only prevents `NaN` predictions from breaking the evaluation metrics, it doesn't solve cold start for genuinely new users or movies.
- **A narrow hyperparameter search.** Only 8 combinations of `rank`, `maxIter`, and `regParam` were tried. That doesn't mean a better configuration doesn't exist, it means the search space covered here was fairly small.
- **Small dataset relative to the technology's intended scale.** The entire point of choosing Spark and ALS is to handle data too large for one machine, but this notebook runs on the same ~100k-rating MovieLens dataset as its SVD companion, which comfortably fits in memory on a laptop. The scalability benefits this architecture is built for are not actually exercised or measured here.
- **R² should be read in context.** Collaborative filtering on explicit ratings tends to produce a modest R², since a lot of the variance in any single rating comes down to individual mood and taste that no amount of cross-user pattern-matching can fully capture. A moderate R² here is expected, not necessarily a sign of a poorly fit model.

## 14. Future Work / Suggestions

- **Actually test at scale.** Run this same pipeline against a substantially larger MovieLens release (1M or 25M ratings) or a synthetic dataset generated to be much larger, ideally on a real multi-node Spark cluster or a cloud-managed Spark service, to see whether the scalability argument for choosing ALS over a single-machine library holds up in practice rather than just in theory.
- **Widen the hyperparameter search.** Trying a wider range of `rank` and `regParam` values, or switching to Spark's `TrainValidationSplit` for a faster (if less thorough) search, could reveal whether a better configuration exists outside the range already tested.
- **Hybrid or content-based signals for cold start.** `genres.csv` and `tags.csv` are already being loaded. Feeding that metadata into the model, or blending it with a content-based similarity score, would give the system something to fall back on for new movies and new users that pure collaborative filtering can't score today.
- **Benchmark directly against the SVD companion notebook** on identical train/test splits, to get an apples-to-apples comparison of accuracy rather than assuming the two approaches perform similarly.
- **Wrap `get_top_n_recommendations()` in a small API or Streamlit demo** so recommendations can be queried interactively instead of by editing and rerunning notebook cells.
- **Incorporate implicit feedback signals**, like tagging activity or watch counts, alongside the explicit star ratings for a potentially richer training signal.
- **Clean up the duplicated rating distribution cell** and the unused template imports noted above, to keep the notebook itself as tidy as the analysis it presents.
